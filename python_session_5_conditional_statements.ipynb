```python
#Im entering to bus 
#conductor is asking how much distance is travel
#if the distance more than 25k then i need to pay charge
#otherwise free ride
distance=eval(input('enter a distance:'))#100
if distance>=25: #100>=25
    charge=eval(input('enter the charge :'))# 2
    charg_dist=distance-25#100-25=75
    total_charge=charg_dist*charge
    print(f'The total charge is:{total_charge}')
else:
    print('enjoy the free ride')
```

**if_elif_else**

- when only **if** is available: one answer

- when **if-else** is available : one condition two possible

- when **if-elif-else** is available : two conditions,3 answers are possible

```python
if con1:
    ans1
elif con2:
    ans2
else:
    ans3
```
