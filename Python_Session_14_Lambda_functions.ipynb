**lambda functions**

- lambda functions is a function in a single line

- many arguments but only one expression

```python
lambda <arguments>:<expression>
```

```python
def add(a):
    print(a+10)

# argument : a  expression a+10
add(100)
```

```python
add1 =lambda a : a+10
add1(20)
```

```python
def cube(x):
    print(x**3)

def avg(a,b,c):
    print((a+b+c)/3)

def avg1(a,b,c=50):
    print((a+b+c)/3)
```

**if -else**

```python
def even_odd(a):
    if a%2==0:
        print('even')
    else:
        print('odd')
```

```python
lambda  <argument> : <if_out> <if_con> else <else_out>
```

```python
even_odd=lambda a: 'even' if a%2==0 else 'odd'
even_odd(20)
```

```python
lambda a: 'even' if a%2==0
```

```python
def max(a,b):
    if a>b:
        print(a)
    else:
        print(b)
```

```python
max1=lambda a,b:a if a>b else b
max1(20,30)
```
