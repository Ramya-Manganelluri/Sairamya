**$case-3: range(start,stop,step): range(2,20.2)$**

- start = start

- positive step means +ve direction step=2

- last=stop-1

** range(-2,-20,-2)**

- start = start

- negative step means -ve direction step=2

- last = stop+1

```python
for i in range(2,20,2): #possible
    print(i)

for i in range(2,20,-2): #no ans
    print(i)

for i in range(-2,-20,2): # no answer
    print(i)

for i in range(-2,-20,-2):#possible
    print(i)
```
